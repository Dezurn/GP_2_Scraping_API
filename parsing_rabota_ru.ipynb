```python
import requests
from bs4 import BeautifulSoup
```

```python
url_main = 'https://www.rabota.ru/?page=1'
page_main = requests.get(url_main) 
```

```python
page_main
```

```python
soup_main = BeautifulSoup(page_main.text, 'html')
```

```python
print(soup_main.prettify())
```

```python
vacancy_links = []

for link in soup_main.find_all('a'):
        part = link.get('href')
        if part and 'vacancy' in part and '?' in part:
            vacancy_links.append(f"https://www.rabota.ru{part.split('?')[0]}")

vacancy_links
```

```python
vacancy_links[9]
```

```python
url_vacancy = vacancy_links[9]
page_vacancy = requests.get(url_vacancy) 
```

```python
page_vacancy
```

```python
soup_vacancy = BeautifulSoup(page_vacancy.text, 'html')
print(soup_vacancy.prettify())
```

```python
vacancy_info = {}
```

```python
vacancy_info['Название вакансии'] = soup_vacancy.find('h1', {'class' : 'vacancy-card__title'}, itemprop = 'title').text.strip()
```

```python
vacancy_info
```

```python
vacancy_info['Зарплата'] = soup_vacancy.find('h3', {'class' : 'vacancy-card__salary'}).text.replace('\xa0', '')
```

```python
vacancy_info
```

```python
vacancy_info['Требуемые новыки'] = soup_vacancy.find('div', {'class' : 'vacancy-card__skills-list'}).text.split('\n')
```

```python
skills = []

for skill in soup_vacancy.find_all('div', {'class' : 'vacancy-card__skills-item'}):
    skills.append(skill.text.strip())

vacancy_info['Требуемые новыки'] = skills
```
