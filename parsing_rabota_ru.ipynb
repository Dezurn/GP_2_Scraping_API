```python
import requests
from bs4 import BeautifulSoup
```

```python
url_main = 'https://www.rabota.ru/vacancy/43993538/'
page_main = requests.get(url_main) 
```

```python
page_main
```

```python
soup_main = BeautifulSoup(page_main.text, 'html')
```

```python
print(soup_main.prettify())
```

```python
links = []

for link in soup_main.find_all('a'):
    if link.get('href') and 'vacancy' in link.get('href') and '?' in link.get('href'):
        links.append(link.get('href').split('?')[0])

links
```
