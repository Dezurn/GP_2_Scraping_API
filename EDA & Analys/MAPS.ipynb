Нужно избавиться от "Петровский парк," и тд, будем убирать запятые и потом смотреть, что получится

Для этого можно написать функцию которая будет идти по элементам списка в столбце и убирать ','

И будем убирать дубликаты

```python
def clearing(i):
  new_i = list()
  i1 = set(i)
  i2 = list(i1)
  for j in i2:
    j = j.replace(',', '')
    j = j.replace(' ', '')
    new_i.append(j)
  return new_i
```

```python
df['Метро'] = df['Метро'].apply(clearing)
```

```python
categorical_subset = (df['Метро'].explode())
categorical_subset = pd.get_dummies(categorical_subset)

categorical_subset = pd.get_dummies(categorical_subset).groupby(level=0).max()
features = pd.concat([df, categorical_subset], axis=1)

stations_list = categorical_subset.columns

counts = {}

for station in stations_list:
    counts[station] = features[station].sum()
counts_df = pd.DataFrame(counts.items(), columns=['station', 'count'])
counts_df.columns = ['Russian_name', 'count']

metro_stations['Russian_name'] = metro_stations['Russian_name'].str.strip().str.lower().str.replace(' ', '')
counts_df['Russian_name'] = counts_df['Russian_name'].str.strip().str.lower()

merged_counts = counts_df.merge(metro_stations[['Russian_name', 'Lat', 'Lon']], on='Russian_name', how='left')
merged_counts.drop_duplicates()

merged_counts = merged_counts.drop_duplicates(subset=['Russian_name'])
merged_counts = merged_counts.dropna()

merged_counts.sort_values(by='count', ascending=False)
```

```python
import folium
from folium.plugins import MarkerCluster
```

```python
k1 = (58 + 1) / 3
k2 = (58 + 1) / 3 * 2
k3 = (58 + 1)
```

```python
def colors(k):
  if k < k1:
    return 'green'
  if k < k2:
    return 'yellow'
  if k < k3:
    return 'orange'
  else:
    return 'red'
```
