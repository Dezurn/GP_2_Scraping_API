Запарсить итоговый датасет через Api у нас получилось 4 отдельными блоками данных. Поэтому их нужно сначала объединить

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dfs = [pd.read_csv('../datasets/superjob_vacancies.csv'), pd.read_csv('../datasets/superjob_vacancies_part_2.csv'), pd.read_csv('../datasets/superjob_vacancies_moscow_part1.csv'), pd.read_csv('../datasets/superjob_vacancies_moscow_part2.csv')]
```

```python
df_sj = pd.concat(dfs, ignore_index=True)
df_sj
```

```python
df_sj.to_csv('../datasets/superjob_all.csv', index=False)
```
