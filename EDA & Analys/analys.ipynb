```python
df_sj, df_rr = pd.read_csv('../datasets/superjob_formated.csv'), pd.read_csv('../datasets/rabota_ru_final.csv', index_col=0)
df_sj
```

Теперь нужно объединить датасеты. Будем считать датасет rabota_ru главным, так как в нем больше данных. Нужно оставить общие колонки у двух датасетов (уникальные признаки мы посмотрели для каждого датасета в соответствующих EDA). Названия колонок в итоговом датасете: Название вакансии, Минимальная зарплата, Максимальная зарплата, Опыт работы, Образование, Компания(нужно изменить названия колонки во 2м датасете), Город, Адрес, Широта, Долгота, Метро, ссылка.

```python

df_rr_s = df_rr[['Название вакансии', 'Минимальная зарплата', 'Максимальная зарплата', 
                        'Опыт работы', 'Образование', 'Компания', 'Город', 'Адрес', 
                        'Широта', 'Долгота', 'Метро', 'Ссылка']].copy()
df_rr_s['Источник'] = 'rabota.ru'
df_sj_s = df_sj[['Название вакансии', 'Минимальная зарплата', 'Максимальная зарплата', 
                        'Опыт', 'Образование', 'Название компании', 'Город', 'Адрес', 
                        'Широта', 'Долгота', 'Метро', 'ссылка']].copy()


df_sj_s = df_sj_s.rename(columns={'Опыт': 'Опыт работы', 'Название компании': 'Компания', 'ссылка': 'Ссылка'})
df_sj_s['Источник'] = 'superjob.ru'


df_ult= pd.concat([df_rr_s, df_sj_s], ignore_index=True)
df_ult
```

В некоторых колонках одни и те же значения написаны по-разному (например с заглавной буквы, "не имеет значения" и "любое"). нужно привести все значения к единому формату

```python
df_ult['Опыт работы'].value_counts()
```

```python
df_ult['Опыт работы'] = df_ult['Опыт работы'].str.lower().str.strip()
df_ult.loc[df_ult['Опыт работы'] == 'не имеет значения', 'Опыт работы'] = 'без опыта'
df_ult.loc[df_ult['Опыт работы'] == 'менее года', 'Опыт работы'] = 'до года'
df_ult['Опыт работы'].value_counts()
```

```python
df_ult['Образование'].value_counts()
```

```python
df_ult['Образование'] = df_ult['Образование'].str.lower().str.strip()
df_ult.loc[df_ult['Образование'] == 'не имеет значения', 'Образование'] = 'любое'
df_ult.loc[df_ult['Образование'] == 'среднее специальное', 'Образование'] = 'среднее профессиональное' #вроде бы это одно и то же, в разных источниках разные названия
df_ult['Образование'].value_counts()
```

```python
df_ult['Средняя зарплата'] = (df_ult['Минимальная зарплата'] + df_ult['Максимальная зарплата']) / 2
df_ult
```

```python
df_ult['Название вакансии'] = df_ult['Название вакансии'].str.lower().str.strip() #также уберет немного разночтения
```

```python
df_ult.to_csv('../datasets/all_vacancies.csv', index=False)
```

Посмотрим на распределение зарплат в зависимости от источника вакансий.

```python
df_sal = df_ult.dropna(subset=['Средняя зарплата'])
fig = plt.figure(figsize=(10, 6))
sns.boxplot(x='Источник', y='Средняя зарплата', data=df_sal)

plt.title('Распределение средней зарплаты по источникам')
plt.show()
```

Видим, что распределение вакансий по зарплатам очень похоже. Это говорит о том, что на обоих площадках схожи запросы работодателей и их вакансий. В  rabota.ru намного больше выброс, так как в нем больше данных.

Теперь посмотрим на самые популярные вакансии на сайтах

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
rabota_top = df_ult[df_ult['Источник'] == 'rabota.ru']['Название вакансии'].value_counts().head(10)

superjob_top = df_ult[df_ult['Источник'] == 'superjob.ru']['Название вакансии'].value_counts().head(10)

axes[0].barh(rabota_top.index, rabota_top.values, color='purple')
axes[0].set_title('Топ-10 самых популярных вакансий rabota.ru')
axes[0].set_xlabel('Количество вакансий')
axes[0].invert_xaxis()
axes[0].invert_yaxis()

axes[1].barh(superjob_top.index, superjob_top.values, color='orange')
axes[1].set_title('Топ-10 самых популярных вакансий superjob.ru')
axes[1].set_xlabel('Количество вакансий')
axes[1].yaxis_inverted()
axes[1].yaxis.tick_right() 
axes[1].invert_yaxis()

plt.show()
```

Видны общие популярные вакансии: продавцы, повара/пекари, курьеры. Но при этом много вакансий уникальных, что говорит о различии в работодателях на площадках. Также выделяются 3 вакансии от магнита на superjob. Посмотрим на компании более детально

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
rabota_top = df_ult[df_ult['Источник'] == 'rabota.ru']['Компания'].value_counts().head(10)

superjob_top = df_ult[df_ult['Источник'] == 'superjob.ru']['Компания'].value_counts().head(10)

axes[0].barh(rabota_top.index, rabota_top.values, color='purple')
axes[0].set_title('Топ-10 самых популярных компаний на rabota.ru')
axes[0].set_xlabel('Количество вакансий')
axes[0].invert_xaxis()
axes[0].invert_yaxis()

axes[1].barh(superjob_top.index, superjob_top.values, color='orange')
axes[1].set_title('Топ-10 самых популярных компаний на superjob.ru')
axes[1].set_xlabel('Количество вакансий')
axes[1].yaxis_inverted()
axes[1].yaxis.tick_right() 
axes[1].invert_yaxis()

plt.show()
```

И тут опять мы видим большие различия: на rabota.ru вакансии размещают различиные компании (максимум всего 16), тогда как на superjob есть крупные поставщики (400 вакансий).

Много вакансий в rabota размещают доставки (Самокат и Яндекс.Доставка), при этом на superjob доставок в топе нет вообще.В superjob топ3 компаний занимают продуктовые сети (Пятерочка, Магнит, Перекресток), а в rabota такой категории даже нет. 

Но есть и пересечения: jobcart и jobers. Это компании которые размещают различные вакансии для других работодателей, а значит их лучше не учитывать при анализе. 

Многие компании являются крупными игроками (помимо уже перечисленных: Билайн, СБЕР, РЖД, Почта России), но на superjob их присутствие больше.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

gr1 = df_ult.groupby(['Источник', 'Опыт работы']).size().reset_index(name='количество')
#нужны относительные значения, иначе график будет неинформативным из-за разницы в количестве данных
gr1.loc[gr1['Источник'] == 'rabota.ru', 'количество%'] = gr1.loc[gr1['Источник'] == 'rabota.ru', 'количество'] / len(df_ult[df_ult['Источник'] == 'rabota.ru']) * 100
gr1.loc[gr1['Источник'] == 'superjob.ru', 'количество%'] = gr1.loc[gr1['Источник'] == 'superjob.ru', 'количество'] / len(df_ult[df_ult['Источник'] == 'superjob.ru']) * 100
sns.barplot(x='Опыт работы', y='количество%', hue='Источник', data=gr1, ax=axes[0])
axes[0].set_title('Распределение вакансий по опыту работы')
axes[0].set_xlabel('Опыт работы')
axes[0].set_ylabel('Количество вакансий (%)')
axes[0].tick_params(axis='x', rotation=45)

gr2 = df_ult.groupby(['Источник', 'Опыт работы'])['Средняя зарплата'].median().reset_index(name='Средняя зарплата')
sns.barplot(x='Опыт работы', y='Средняя зарплата', hue='Источник', data=gr2, ax=axes[1])
axes[1].set_title('Средняя зарплата по опыту работы')
axes[1].set_xlabel('Опыт работы')
axes[1].set_ylabel('Средняя зарплата')
axes[1].tick_params(axis='x', rotation=45)

plt.show()
```

Большинство вакансий на обоих площадках не требуют какого либо опыта, при этом интересно, что топ3 самого часто требуемого опыта также совпадает.

На rabota.ru представлено большее количество различных категорий опытов работы. В 3х категориях данные по зарплате присутствуют одновременно в обоих датасетах ,и во всех них средняя зарплата на обоих площадках оценивается примерно одинаково. Это говорит о схожести рынков труда на обоих сайтах.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

gr1 = df_ult.groupby(['Источник', 'Образование']).size().reset_index(name='количество')
gr1.loc[gr1['Источник'] == 'rabota.ru', 'количество%'] = gr1.loc[gr1['Источник'] == 'rabota.ru', 'количество'] / len(df_ult[df_ult['Источник'] == 'rabota.ru']) * 100
gr1.loc[gr1['Источник'] == 'superjob.ru', 'количество%'] = gr1.loc[gr1['Источник'] == 'superjob.ru', 'количество'] / len(df_ult[df_ult['Источник'] == 'superjob.ru']) * 100
sns.barplot(x='Образование', y='количество%', hue='Источник', data=gr1, ax=axes[0])
axes[0].set_title('Распределение вакансий по образованию')
axes[0].set_xlabel('Образование')
axes[0].set_ylabel('Количество вакансий (%)')
axes[0].tick_params(axis='x', rotation=45)

gr2 = df_ult.groupby(['Источник', 'Образование'])['Средняя зарплата'].median().reset_index(name='Средняя зарплата')
sns.barplot(x='Образование', y='Средняя зарплата', hue='Источник', data=gr2, ax=axes[1])
axes[1].set_title('Средняя зарплата по образованию')
axes[1].set_xlabel('Образование')
axes[1].set_ylabel('Средняя зарплата')
axes[1].tick_params(axis='x', rotation=45)

plt.show()
```

Решил также посмотреть на распределение по требуемому образованию, и увидел практически те же зависимости, только еще нагляднее. В обоих датасетах есть данные по всем категориям, и при этом топы по полуярности также совпадают. Среднии зарплаты немного расходятся, особенно в категории "среднее", но тут проблема в том, что все зарплаты находятся примерно в одном диапазоне 70-80к. В целом зарплаты на superjob немного ниже чем на rabota.ru.
