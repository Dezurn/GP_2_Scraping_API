Запарсить итоговый датасет через Api у нас получилось 4 отдельными блоками данных. Поэтому их нужно сначала объединить

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dfs = [pd.read_csv('superjob_vacancies.csv'), pd.read_csv('superjob_vacancies_part_2.csv'), pd.read_csv('superjob_vacancies_moscow_part1.csv'), pd.read_csv('superjob_vacancies_moscow_part2.csv')]
```

```python
df_sj = pd.concat(dfs, ignore_index=True)
df_sj
```

```python
df_sj.to_csv('superjob_all.csv', index=False)
```

Из-за такого сбора данных могло получится много дубликатов. Посмотрим на них:

```python
df_sj[df_sj.duplicated(keep='first')]
```

```python
df_sj = df_sj.drop_duplicates()
df_sj
```

Посмотрим кол-во пропусков в данных

---



```python
(df_sj.isna().mean() * 100).round(2).sort_values(ascending=False)
```

```python
df_sj = df_sj.drop(columns=['Индуcтрия'])
```

Удалим признак индустрия, т.к он полностью пустой. Видим, что, если у нас отсутствует Название компании, то и отсутствует Кол-во персонала в компании и Кол-во размещенных вакансий. Тоже самое и с Адресом, если его нет, то нету и Широты, Долготы. Удалять эти строки мы не будем, т.к потеряем большой кусок данных для анализа, где не нужен точный адрес компании. Признак Название компании также не будем удалять или искусствено заполнять, потому что пропусков там мало и они тоже могут нести смысловую информацию, например часть компаний по каким-то причинам решили не раскрывать свое название.

Дальше посмотрим на скрытые пропуски, они нас интересуют в большей степени.

```python
print((df_sj['Минимальная зарплата'] == 0).sum(), (df_sj['Минимальная зарплата'] == 0).mean() * 100)
print((df_sj['Максимальная зарплата'] == 0).sum(), (df_sj['Максимальная зарплата'] == 0).mean() * 100)
print(((df_sj['Минимальная зарплата'] == 0) & (df_sj['Максимальная зарплата'] == 0)).sum(), ((df_sj['Минимальная зарплата'] == 0) & (df_sj['Максимальная зарплата'] == 0)).mean() * 100)
```

Удалим все строки, где в признаках зарплаты есть 0, это не настоящие 0, а это значит, что минимальная или максимальная зарплата не указана. Инфо о вакансии без зарплаты нам не нужны.

```python
df_sj = df_sj.drop(df_sj[(df_sj['Минимальная зарплата'] == 0)].index)
df_sj = df_sj.drop(df_sj[(df_sj['Максимальная зарплата'] == 0)].index)
```

```python
print((df_sj['Метро'] == '[]').sum())
print((df_sj['Категория'] == '[]').sum())
print((df_sj['Детальные категории'] == '[]').sum())
```

В признаке Категория нет пустых списков, то есть пропусков. Это хорошо, то что в признаке метро есть небольшая доля пропусков ничего страшного, есть вакансии, где нет смысла указывать метро или его просто нет рядом.

```python
df_sj['Средняя зарплата'] = df_sj[['Минимальная зарплата', 'Максимальная зарплата']].mean(axis=1)
```

Создадим новый признак средняя зарплата

```python
df_sj[['Минимальная зарплата', 'Максимальная зарплата', 'Средняя зарплата']].describe()
```

```python
sns.boxplot(x=df_sj['Средняя зарплата'])
plt.title('Анализ выбросов на Boxplot')
```

Видно, что есть пару бросающихся выбросов, посмотрим на них

```python
df_sj.sort_values('Средняя зарплата', ascending=False).head(20)
```

```python
df_sj.sort_values('Средняя зарплата').head(20)
```

Есть пару вакансий с неадекватными зарплатами, удалим их, чтобы не мешали дальнейшему анализу

```python
df_sj = df_sj[(df_sj['Средняя зарплата'] >= 2000) & (df_sj['Средняя зарплата'] <= 500000)]
```

Посмотрим на гистограмму средней зарплаты

```python
df_sj['Средняя зарплата'].hist(grid=False, edgecolor='black')
plt.title('Распределение средней зарплаты')
plt.xlabel('Средняя зарплата')
plt.ylabel('Количество вакансий')
```

Видно, что средняя зепка у большинства вакансий около 80к, посмотрим в каких городах больше всего вакансий.

```python
df_sj['Город'].value_counts().head(15)
```

Неудевительно, что в больших горадах больше всего и вакансий на работу, дальше посмотрим на графики зависимости количетсва вакансий от опыта и образования

```python
df_sj['Опыт'].value_counts().plot.bar()
plt.title('Распределение вакансий по опыту')
plt.xlabel('Опыт')
plt.ylabel('Количество вакансий')
```

```python
df_sj['Образование'].value_counts().plot.bar()
plt.title('Распределение вакансий по образованию')
plt.xlabel('Образование')
plt.ylabel('Количество вакансий')
```

Больше всего вакансий на людей с минимальным опытом и без опыта + образование их не имеет значения или устраивает среднее. Теперь нужно посмотреть на зависимость зарплаты и опыта и образлвания.

```python
salary_exp = df_sj.groupby('Опыт')['Средняя зарплата'].agg(['median']).sort_values('median', ascending=False)
salary_exp['median'].plot.bar()
plt.xlabel('Опыт')
plt.ylabel('Зарплата')
```

```python
salary_edu = df_sj.groupby('Образование')['Средняя зарплата'].agg(['median']).sort_values('median', ascending=False)
salary_edu['median'].plot.bar()
plt.xlabel('Образование')
plt.ylabel('Зарплата')
```

Вакансии, где требуется большой опыт работы и образолвание - самые высоко оплачивыемые. Так что можно точно сказать, что зарплата напрямую зависит от опыта работы и образования.
